# Training and Testing a Model

## Load Necessary Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv("SPY_2000_2022.csv")
df
```

```python
df.drop(columns= "Date")
```

## spliting the data for training and testing

```python
# Giving the X parameters and y parameters
X = df[["High", "Low", "Open", "Volume"]]
y = df["Adj Close"]
```

```python
X_train, X_test, y_train, y_test  = train_test_split(X, y, 
                                                     test_size=0.2,
                                                     shuffle=False, 
                                                     random_state=35,)
```

## Fitting the model

```python
model = LinearRegression()
model.fit(X_train, y_train)

```

_extracting slope and intercept_

```python
slope = model.coef_
intercept = model.intercept_


print(f"Slope: {slope}")
print(f"Intercept: {intercept}")
```

## The estimated Regression Model is given by: $\hat{y} = \omega_0 + \omega_1 * X$
- where $\omega_0$ = bias = -0.012968817013259581  
- And $\omega_1$ = $X1$ =  8.09337281e-01  
- And $\omega_2$ = $X2$ = 8.00855110e-01  
- And $\omega_3$ = $X3$ = -6.09656651e-01   
- And $\omega_4$ = $X4$ = 1.47180173e-11

## Making Prediction


```python
y_pred = model.predict(X_test)

data = {"actual": y_test, "Predicted": y_pred}

# Turning the Dictionary to a pandas dataframe
pred_df = pd.DataFrame(data)
pred_df
```

## Performance of the Model on the test set

```python
mse = mean_squared_error(y_pred, y_test)
rmse = np.sqrt(mse)
r2 = r2_score(y_pred, y_test)

print(f"MSE  = {mse:.4f}")
print(f"RMSE = {rmse:.4f}")
print(f"R²   = {r2:.4f}")
```

## Performance of the Model on the Training set

```python
y_pred_train = model.predict(X_train)

mse_train = mean_squared_error(y_pred_train, y_train)
rmse_train = np.sqrt(mse_train)
r2_train = r2_score(y_pred_train, y_train)

print(f"MSE  = {mse_train:.4f}")
print(f"RMSE = {rmse_train:.4f}")
print(f"R²   = {r2_train:.4f}")
```
